# **Dataset: Student Alcohol Consumption**

Os dados foram obtidos por meio de uma pesquisa com estudantes do ensino secundário que cursavam Matemática e Língua Portuguesa. O conjunto de dados contém diversas **informações sobre os aspectos sociais, o gênero e os hábitos de estudo dos estudantes.**

# **Dicionário**

Atributos presentes nos dois conjuntos de dados: `student-mat.csv` (curso de Matemática) e `student-por.csv` (curso de Língua Portuguesa).

1. sex - student's sex (binary: 'F' - female or 'M' - male)
2. age - student's age (numeric: from 15 to 22)
3. address - student's home address type (binary: 'U' - urban or 'R' - rural)
4. famsize - family size (binary: 'LE3' - less or equal to 3 or 'GT3' - greater than 3)
5. Pstatus - parent's cohabitation status (binary: 'T' - living together or 'A' - apart)
6. Medu - mother's education (numeric: 0 - none, 1 - primary education (4th grade), 2 – 5th to 9th grade, 3 – secondary education or 4 – higher education)
7. Fedu - father's education (numeric: 0 - none, 1 - primary education (4th grade), 2 – 5th to 9th grade, 3 – secondary education or 4 – higher education)
8. Mjob - mother's job (nominal: 'teacher', 'health' care related, civil 'services' (e.g. administrative or police), 'at_home' or 'other')
9. Fjob - father's job (nominal: 'teacher', 'health' care related, civil 'services' (e.g. administrative or police), 'at_home' or 'other')
10. reason - reason to choose this school (nominal: close to 'home', school 'reputation', 'course' preference or 'other')
11. guardian - student's guardian (nominal: 'mother', 'father' or 'other')
12. traveltime - home to school travel time (numeric: 1 - <15 min., 2 - 15 to 30 min., 3 - 30 min. to 1 hour, or 4 - >1 hour)
13. studytime - weekly study time (numeric: 1 - <2 hours, 2 - 2 to 5 hours, 3 - 5 to 10 hours, or 4 - >10 hours)
14. failures - number of past class failures (numeric: n if 1<=n<3, else 4)
15. schoolsup - extra educational support (binary: yes or no)
16. famsup - family educational support (binary: yes or no)
17. paid - extra paid classes within the course subject (Math or Portuguese) (binary: yes or no)
18. activities - extra-curricular activities (binary: yes or no)
19. nursery - attended nursery school (binary: yes or no)
20. higher - wants to take higher education (binary: yes or no)
21. internet - Internet access at home (binary: yes or no)
22. romantic - with a romantic relationship (binary: yes or no)
23. famrel - quality of family relationships (numeric: from 1 - very bad to 5 - excellent)
24. freetime - free time after school (numeric: from 1 - very low to 5 - very high)
25. goout - going out with friends (numeric: from 1 - very low to 5 - very high)
26. Dalc - workday alcohol consumption (numeric: from 1 - very low to 5 - very high)
27. Walc - weekend alcohol consumption (numeric: from 1 - very low to 5 - very high)
28. health - current health status (numeric: from 1 - very bad to 5 - very good)
29. absences - number of school absences (numeric: from 0 to 93)

21. G1 — Nota do primeiro período (numérica: de 0 a 20).

22. G2 — Nota do segundo período (numérica: de 0 a 20).

23. G3 — Nota final (numérica: de 0 a 20; variável-alvo para previsão).

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
import pandas as pd
d1 = pd.read_csv('student-por.csv')
d1 = d1.drop(columns=['school'], axis=1)
display(d1.head())
print('Dimensoes:', d1.shape)
```

```python
d1 = pd.read_csv('student-por.csv')
d1 = d1.drop(columns=['school'], axis=1)
display(d1.head())
print("Dimensões:", d1.shape)
```

```python
d1 = d1.rename(columns={
    'sex': 'sexo',
    'age': 'idade',
    'address':  'regiao',
    'absences': 'faltas',
    'famsize': 'tamanho_familia',
    'Pstatus': 'convivencia_pais',
    'guardian': 'guardiao_legal',
    'studytime': 'tempo_semanal_estudo',
    'failures': 'nº_reprovacoes',
    'famsup': 'apoio_familiar',
    'activities': 'participacao_atividades_extras',
    'higher': 'desejo_de_ensino_superior',
    'internet': 'acesso_internet_casa',
    'romantic': 'relacionamento_amoroso',
    'famrel': 'qualidade_relacao_familiar',
    'freetime': 'tempo_livre',
    'goout': 'frequencia_saida_com_amigos',
    'Dalc': 'consumo_alcool_semana',
    'Walc': 'consumo_alcool_fim_de_semana',
    'health': 'estado_saude_atual',
    'G1': 'nota_1p',
    'G2': 'nota_2p',
    'G3': 'nota_final',
    'Medu': 'educacao_mae',
    'Fedu': 'educacao_pai',
    'Mjob': 'trabalho_mae',
    'Fjob': 'trabalho_pai',
    'reason': 'motivo_escolha_escola',
    'traveltime': 'tempo_deslocamento',
    'schoolsup': 'apoio_escolar',
    'paid': 'aulas_extras_pagas',
    'nursery': 'frequentou_creche'
})
```

```python
# adicionar coluna com a média das 3 notas de cada aluno
d1["media_notas"] = d1[["nota_1p", "nota_2p", "nota_final"]].mean(axis=1).round(1)
display(d1.head())
```

```python
print(d1.isnull().sum())
# não tem nulos
```

```python
print("Duplicatas:", d1.duplicated().sum())
# não tem duplicatas
```

```python
#criando os grupos q1 e q4
q1 = d1[d1['media_notas'] <= d1['media_notas'].quantile(0.25)].copy()
q4 = d1[d1['media_notas'] >= d1['media_notas'].quantile(0.75)].copy()
```

# RELATÓRIO - Análise de Barreira (Cláusula > 0.5 ponto)

**Critério:** Se a diferença absoluta entre as médias de `media_notas` dos dois grupos binários for > 0.5, a variável **PASSA** pela cláusula de barreira. Caso contrário (**≤ 0.5**), **NÃO PASSA**.

**Variáveis analisadas:** sexo, regiao, tamanho_familia, convivencia_pais, apoio_escolar, apoio_familiar,
            aulas_extras_pagas, participacao_atividades_extras, frequentou_creche, desejo_de_ensino_superior,
            acesso_internet_casa, relacionamento_amoroso.

Os resultados detalhados com médias, diferenças e status encontram-se na célula abaixo.


```python
# Variáveis binárias identificadas
binarias = ['sexo', 'regiao', 'tamanho_familia', 'convivencia_pais', 'apoio_escolar', 'apoio_familiar',
            'aulas_extras_pagas', 'participacao_atividades_extras', 'frequentou_creche', 'desejo_de_ensino_superior',
            'acesso_internet_casa', 'relacionamento_amoroso']

resultados = []

for var in binarias:
    if var in d1.columns:
        grupos = d1.groupby(var)['media_notas'].agg(['mean', 'count'])
        valores = grupos['mean'].values
        if len(valores) == 2:
            media1, media2 = valores[0], valores[1]
            diff = abs(media1 - media2)
            passou = diff > 0.5
            cat1, cat2 = grupos.index[0], grupos.index[1]
            resultados.append({
                'variáveis': var,
                'diferença absoluta': round(diff, 3),
                'passou > 0.5': 'SIM' if passou else 'NÃO'
            })
            print(f"{var}: {cat1}={media1:.3f}, {cat2}={media2:.3f}, |diff|={diff:.3f} -> {'PASSOU' if passou else 'NÃO PASSOU'}")
    else:
        print(f"AVISO: {var} não encontrada no dataframe")

print("\n" + "="*60)
df_resultado = pd.DataFrame(resultados)
df_resultado = df_resultado[['variáveis', 'diferença absoluta', 'passou > 0.5']]
display(df_resultado)
```

```python
#eliminando as variáveis
nao_passou = ['tamanho_familia', 'convivencia_pais', 'apoio_familiar', 'participacao_atividades_extras', 'frequentou_creche']
d1 = d1.drop(columns=nao_passou, axis=1)
```

```python
# GRÁFICOS PIZZA - VARIÁVEIS BINÁRIAS/CATEGÓRICAS (Q1 vs Q4) - CONSOLIDADOS
# Cores padronizadas com mapeamento consistente: salmon / skyblue

vars_pizza = [
    ('regiao', 'Região'),
    ('sexo', 'Sexo'),
    ('desejo_de_ensino_superior', 'Desejo de Ensino Superior'),
    ('relacionamento_amoroso', 'Relacionamento Amoroso'),
    ('apoio_escolar', 'Apoio Escolar'),
    ('aulas_extras_pagas', 'Aulas Extras Pagas'),
    ('acesso_internet_casa', 'Acesso Internet Casa'),
]

fig, axes = plt.subplots(len(vars_pizza), 2, figsize=(14, 5*len(vars_pizza)))

for i, (var, titulo) in enumerate(vars_pizza):
    if var not in q1.columns or var not in q4.columns:
        continue

    q1_counts = q1[var].value_counts()
    q4_counts = q4[var].value_counts()

    # Mapeamento consistente de cores por categoria
    todas_cats = list(q1_counts.index) + list(q4_counts.index)
    # Define cores padronizadas
    cat_color = {}
    for j, cat in enumerate(sorted(set(todas_cats))):
        if j == 0:
            cat_color[cat] = 'salmon'
        else:
            cat_color[cat] = 'skyblue'

    cores_q1 = [cat_color.get(c, 'gray') for c in q1_counts.index]
    cores_q4 = [cat_color.get(c, 'gray') for c in q4_counts.index]

    axes[i, 0].pie(q1_counts, labels=q1_counts.index, colors=cores_q1, autopct='%1.1f%%', startangle=90)
    axes[i, 0].set_title(f'Q1 - {titulo}')

    axes[i, 1].pie(q4_counts, labels=q4_counts.index, colors=cores_q4, autopct='%1.1f%%', startangle=90)
    axes[i, 1].set_title(f'Q4 - {titulo}')

plt.suptitle('Distribuição por categoria - Comparação Q1 vs Q4', fontsize=16, y=1.001)
plt.tight_layout()
plt.show()
```

# **Conclusões - extremos**

## Principais Observações

| Variável | Observação |
| --- | --- |
| **Região** | O grupo Q4 apresenta maior concentração em áreas urbanas (U), enquanto o grupo Q1 possui maior proporção de alunos da zona rural (R). |
| **Sexo** | No grupo Q1, a distribuição entre meninos (M) e meninas (F) é mais equilibrada, com uma vantagem marginal de meninos. Já no grupo Q4, há uma proporção de dois quartos do sexo feminino. |
| **Desejo de Ensino Superior** | Enquanto aproximadamente um quarto dos alunos do Q1 não demonstra interesse em cursar o ensino superior, no grupo Q4 essa proporção é praticamente nula, indicando forte desejo de continuidade nos estudos. |
| **Relacionamento Amoroso** | O grupo Q4 apresenta um maior percentual de alunos que afirmam **não** possuir relacionamento amoroso em comparação ao grupo Q1, sugerindo que, entre os melhores desempenhos acadêmicos, há uma maior proporção de estudantes sem envolvimento amoroso. |
| **Apoio Escolar** | Observa-se uma redução relativa de 65% de apoio escolar entre o grupo q1 e q4, indicando que alunos com menor rendimento são os mais abonados com counteúdos educacionais complementares |
| **Aulas Extras Pagas** | Há uma redução de quase 50% de alunos que pagam aulas extras pagas entre os grupos, indicando que alunos com pior desempenho são os que mais vão atrás de reforço acadêmico. |
| **Acesso à Internet em Casa** | O grupo Q4 apresenta maior proporção de alunos com acesso à internet em casa, o que pode estar associado a melhores condições para estudo e pesquisa. |

## Considerações Finais

- **Fator socioeconômico:** Variáveis como região e acesso à internet sugerem que melhores condições de infraestrutura estão mais presentes entre os alunos com maiores médias.
- **Projeto de vida:** O desejo de ensino superior mostra-se fortemente associado ao desempenho, sendo praticamente unânime no grupo Q4.
- **Aspectos sociais:** Diferenças nas variáveis de relacionamento e apoio indicam possíveis trocas entre vida social e desempenho acadêmico.
